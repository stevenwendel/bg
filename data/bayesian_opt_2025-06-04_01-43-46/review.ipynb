```python
import pandas as pd
import matplotlib.pyplot as plt
import pickle

pklfile = "/Users/stevenwendel/Documents/GitHub/bg/data/bayesian_opt_2025-06-06_08-37-58/optimization_results.pkl"

# Load the pickle file
with open(pklfile, 'rb') as f:
    results = pickle.load(f)
```

```python
sorted(results, key=lambda x: x['best_score'], reverse=True)
```

```python
scores = [i['best_score'] for i in results]
for param in ['MUT_RATE', 'MUT_SIGMA', 'ELITE_SIZE', 'POP_SIZE']:
    data = [i['parameters'][param] for i in results]
    
    plt.plot(data,scores, 'o')
    plt.title(f'{param}')
    plt.show()       
```
